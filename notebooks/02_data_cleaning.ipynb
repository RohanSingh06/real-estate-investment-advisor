# Phase 3 — Data Cleaning

## Real Estate Investment Advisor

This notebook performs systematic data cleaning on the India housing prices dataset.

The raw dataset is preserved separately. Cleaning operations are applied to a working copy, and the cleaned dataset is saved to the processed data directory.

### Objectives

- Load the raw dataset
- Verify dataset structure
- Handle missing values
- Remove duplicate records if present
- Validate categorical values
- Validate numerical values
- Preserve important source information
- Document known structural inconsistencies
- Save the cleaned dataset

```python
import pandas as pd
import numpy as np
import os

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Libraries imported successfully.")
```

```python
raw_path = "../data/raw/india_housing_prices.csv"

df = pd.read_csv(raw_path)

print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

```python
df.head()
```

## 1. Dataset Structure

Before applying any cleaning operations, the structure and data types of the raw dataset are verified.

```python
print("Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)
```

## 2. Missing Value Check

The dataset is checked for missing values across all columns.

```python
missing_summary = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().sum() / len(df)) * 100
})

missing_summary
```

## 3. Duplicate Record Check

Duplicate rows are identified before cleaning.

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
if duplicate_count > 0:
    df = df.drop_duplicates().copy()
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found. Dataset unchanged.")
```

## 4. Unique ID Validation

The uniqueness of the property ID is verified.

```python
duplicate_ids = df["ID"].duplicated().sum()

print("Duplicate IDs:", duplicate_ids)
print("Unique IDs:", df["ID"].nunique())
print("Total records:", len(df))
```

```python
if duplicate_ids == 0:
    print("All property IDs are unique.")
else:
    print("Duplicate property IDs require investigation.")
```

## 5. Categorical Text Standardization Check

Text-based categorical columns are inspected for leading/trailing spaces and inconsistent formatting.

```python
categorical_columns = [
    "State",
    "City",
    "Locality",
    "Property_Type",
    "Furnished_Status",
    "Public_Transport_Accessibility",
    "Parking_Space",
    "Security",
    "Amenities",
    "Facing",
    "Owner_Type",
    "Availability_Status"
]

for col in categorical_columns:
    leading_trailing_spaces = (
        df[col].astype(str) != df[col].astype(str).str.strip()
    ).sum()

    print(f"{col}: {leading_trailing_spaces} records with leading/trailing spaces")
```

```python
for col in categorical_columns:
    df[col] = df[col].astype(str).str.strip()
```

```python
print("Text columns standardized successfully.")
```

## 6. Categorical Value Validation

Categorical columns are checked against the values observed during dataset inspection.

```python
for col in categorical_columns:
    print(f"\n{col}")
    print(df[col].value_counts())
```

## 7. Numerical Feature Validation

Numerical columns are checked for invalid, negative, zero, or logically impossible values.

```python
numerical_columns = [
    "ID",
    "BHK",
    "Size_in_SqFt",
    "Price_in_Lakhs",
    "Price_per_SqFt",
    "Year_Built",
    "Floor_No",
    "Total_Floors",
    "Age_of_Property",
    "Nearby_Schools",
    "Nearby_Hospitals"
]

print("Numerical columns:")
print(numerical_columns)
```

```python
validation_results = []

for col in numerical_columns:
    validation_results.append({
        "Column": col,
        "Negative_Count": (df[col] < 0).sum(),
        "Zero_Count": (df[col] == 0).sum(),
        "Minimum": df[col].min(),
        "Maximum": df[col].max()
    })

numerical_validation = pd.DataFrame(validation_results)

numerical_validation
```

## 8. Property-Level Logical Validation

Relationships between related numerical variables are checked for logical consistency.

```python
print("BHK <= 0:", (df["BHK"] <= 0).sum())

print("Size <= 0:", (df["Size_in_SqFt"] <= 0).sum())

print("Price <= 0:", (df["Price_in_Lakhs"] <= 0).sum())

print("Total_Floors <= 0:", (df["Total_Floors"] <= 0).sum())

print("Floor_No < 0:", (df["Floor_No"] < 0).sum())

print("Age_of_Property < 0:", (df["Age_of_Property"] < 0).sum())

print("Year_Built outside 1900-2025:",
      ((df["Year_Built"] < 1900) | (df["Year_Built"] > 2025)).sum())
```

```python
calculated_age = 2025 - df["Year_Built"]

age_difference = df["Age_of_Property"] - calculated_age

print("Age inconsistencies:", (age_difference != 0).sum())
print("Maximum absolute difference:", age_difference.abs().max())
```

```python
floor_inconsistency = df["Floor_No"] > df["Total_Floors"]

print("Floor inconsistencies:", floor_inconsistency.sum())
print(
    "Percentage:",
    round(floor_inconsistency.mean() * 100, 2),
    "%"
)
```

## 9. Price per SqFt Validation

The supplied `Price_per_SqFt` field is compared against a value calculated from the property's price and size.

```python
df["Calculated_Price_per_SqFt"] = (
    df["Price_in_Lakhs"] * 100000
) / df["Size_in_SqFt"]

price_difference = (
    df["Price_per_SqFt"] -
    df["Calculated_Price_per_SqFt"]
)

print("Mean absolute difference:",
      price_difference.abs().mean())

print("Maximum absolute difference:",
      price_difference.abs().max())

print("Median calculated price per sqft:",
      df["Calculated_Price_per_SqFt"].median())

print("Median supplied price per sqft:",
      df["Price_per_SqFt"].median())
```

```python
price_ratio = (
    df["Price_per_SqFt"] /
    df["Calculated_Price_per_SqFt"]
)

print("Records where supplied value differs substantially:",
      (price_ratio < 0.5).sum() + (price_ratio > 1.5).sum())
```

```python
df.drop(columns=["Calculated_Price_per_SqFt"], inplace=True)

print("Temporary investigation column removed.")
print("Current dataset shape:", df.shape)
```

## 10. Final Data Cleaning Decisions

The following cleaning decisions are applied:

1. No missing values were found.
2. No duplicate rows were found.
3. All property IDs are unique.
4. Text columns were stripped of leading/trailing whitespace.
5. Valid numerical values are retained.
6. `Age_of_Property` is internally consistent with `Year_Built`.
7. Floor inconsistencies are retained because correcting or removing 46.52% of the records would not be justified from the available data.
8. The original `Price_per_SqFt` column is retained in the cleaned dataset for traceability. A corrected price-per-square-foot feature will be created during Feature Engineering.
9. No statistical outliers are removed because the observed numerical ranges are plausible for the dataset and the IQR analysis did not identify meaningful outliers in the major numerical variables.

```python
# Remove exact duplicate rows if any
df = df.drop_duplicates().copy()

# Remove leading/trailing whitespace from categorical columns
for col in categorical_columns:
    df[col] = df[col].astype(str).str.strip()

print("Basic cleaning completed.")
print("Shape:", df.shape)
```

```python
final_check = pd.DataFrame({
    "Missing_Values": df.isnull().sum(),
    "Unique_Values": df.nunique(),
    "Data_Type": df.dtypes.astype(str)
})

final_check
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nDuplicate rows:", df.duplicated().sum())
print("Duplicate IDs:", df["ID"].duplicated().sum())

print("\nNegative BHK:", (df["BHK"] < 0).sum())
print("Invalid Size:", (df["Size_in_SqFt"] <= 0).sum())
print("Invalid Price:", (df["Price_in_Lakhs"] <= 0).sum())
print("Negative Age:", (df["Age_of_Property"] < 0).sum())
```

```python
processed_path = "../data/processed"

os.makedirs(processed_path, exist_ok=True)

cleaned_file = "../data/processed/india_housing_prices_cleaned.csv"

df.to_csv(cleaned_file, index=False)

print("Cleaned dataset saved successfully.")
print(cleaned_file)
```

```python
import os

print("File exists:", os.path.exists(cleaned_file))

if os.path.exists(cleaned_file):
    print(
        "File size:",
        round(os.path.getsize(cleaned_file) / (1024 * 1024), 2),
        "MB"
    )
```

```python
cleaned_df = pd.read_csv(cleaned_file)

print("Reloaded dataset shape:", cleaned_df.shape)
print("Columns:", len(cleaned_df.columns))
print("Missing values:", cleaned_df.isnull().sum().sum())
print("Duplicate rows:", cleaned_df.duplicated().sum())
```
