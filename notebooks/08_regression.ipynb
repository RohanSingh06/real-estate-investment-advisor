# Phase 9 — Regression

## Predicting Future Property Price

Target: `Future_Price_5Y`

Models:
1. Linear Regression
2. Decision Tree Regressor
3. Random Forest Regressor
4. Extra Trees Regressor
5. XGBoost Regressor

Metrics:
- MAE
- MSE
- RMSE
- R²

Note: `Future_Price_5Y` is generated from current price using a fixed 8% annual growth assumption for 5 years. Very high predictive performance is therefore expected.

```python
import pandas as pd
import numpy as np
import joblib
import os

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from xgboost import XGBRegressor

print("Libraries imported successfully.")
```

```python
# Load regression data

X_train_reg = joblib.load("../models/X_train_reg_processed.pkl")
X_test_reg = joblib.load("../models/X_test_reg_processed.pkl")
y_train_reg = joblib.load("../models/y_train_reg.pkl")
y_test_reg = joblib.load("../models/y_test_reg.pkl")

print("Regression data loaded successfully.")
print("Training features:", X_train_reg.shape)
print("Testing features :", X_test_reg.shape)
print("Training target  :", y_train_reg.shape)
print("Testing target   :", y_test_reg.shape)

if y_train_reg.isna().any() or y_test_reg.isna().any():
    raise ValueError("Regression target contains NaN. Re-run 04_target_creation and 06_ml_preprocessing.")
```

```python
def evaluate_regression_model(model_name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)

    print(model_name)
    print("=" * 40)
    print(f"MAE : {mae:.4f}")
    print(f"MSE : {mse:.4f}")
    print(f"RMSE: {rmse:.4f}")
    print(f"R²  : {r2:.4f}")

    return mae, mse, rmse, r2
```

```python
# Linear Regression

linear_model = LinearRegression()
linear_model.fit(X_train_reg, y_train_reg)

y_pred_linear = linear_model.predict(X_test_reg)

mae_linear, mse_linear, rmse_linear, r2_linear = evaluate_regression_model(
    "Linear Regression",
    y_test_reg,
    y_pred_linear
)
```

```python
# Decision Tree Regressor

decision_tree_reg_model = DecisionTreeRegressor(
    max_depth=15,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=42
)

decision_tree_reg_model.fit(X_train_reg, y_train_reg)

y_pred_dt_reg = decision_tree_reg_model.predict(X_test_reg)

mae_dt_reg, mse_dt_reg, rmse_dt_reg, r2_dt_reg = evaluate_regression_model(
    "Decision Tree",
    y_test_reg,
    y_pred_dt_reg
)
```

```python
# Random Forest Regressor

random_forest_reg_model = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)

random_forest_reg_model.fit(X_train_reg, y_train_reg)

y_pred_rf_reg = random_forest_reg_model.predict(X_test_reg)

mae_rf_reg, mse_rf_reg, rmse_rf_reg, r2_rf_reg = evaluate_regression_model(
    "Random Forest",
    y_test_reg,
    y_pred_rf_reg
)
```

```python
# Extra Trees Regressor

extra_trees_reg_model = ExtraTreesRegressor(
    n_estimators=100,
    max_depth=15,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)

extra_trees_reg_model.fit(X_train_reg, y_train_reg)

y_pred_et_reg = extra_trees_reg_model.predict(X_test_reg)

mae_et_reg, mse_et_reg, rmse_et_reg, r2_et_reg = evaluate_regression_model(
    "Extra Trees",
    y_test_reg,
    y_pred_et_reg
)
```

```python
# XGBoost Regressor

xgb_reg_model = XGBRegressor(
    n_estimators=100,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    eval_metric="rmse",
    random_state=42,
    n_jobs=-1
)

xgb_reg_model.fit(X_train_reg, y_train_reg)

y_pred_xgb_reg = xgb_reg_model.predict(X_test_reg)

mae_xgb_reg, mse_xgb_reg, rmse_xgb_reg, r2_xgb_reg = evaluate_regression_model(
    "XGBoost",
    y_test_reg,
    y_pred_xgb_reg
)
```

```python
# Regression Model Comparison

regression_results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Decision Tree",
        "Random Forest",
        "Extra Trees",
        "XGBoost"
    ],
    "MAE": [
        mae_linear, mae_dt_reg, mae_rf_reg, mae_et_reg, mae_xgb_reg
    ],
    "MSE": [
        mse_linear, mse_dt_reg, mse_rf_reg, mse_et_reg, mse_xgb_reg
    ],
    "RMSE": [
        rmse_linear, rmse_dt_reg, rmse_rf_reg, rmse_et_reg, rmse_xgb_reg
    ],
    "R2": [
        r2_linear, r2_dt_reg, r2_rf_reg, r2_et_reg, r2_xgb_reg
    ]
})

regression_results
```

```python
# Save regression results

os.makedirs("../models", exist_ok=True)

regression_results.to_csv(
    "../models/regression_results.csv",
    index=False
)

print("Regression results saved successfully.")
```

```python
# Save all regression models

joblib.dump(linear_model, "../models/linear_regression.pkl")
joblib.dump(decision_tree_reg_model, "../models/decision_tree_regressor.pkl")
joblib.dump(random_forest_reg_model, "../models/random_forest_regressor.pkl")
joblib.dump(extra_trees_reg_model, "../models/extra_trees_regressor.pkl")
joblib.dump(xgb_reg_model, "../models/xgboost_regressor.pkl")

print("All regression models saved successfully.")
```
