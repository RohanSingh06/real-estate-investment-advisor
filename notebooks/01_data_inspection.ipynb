## 15. Numerical Outlier Investigation

Before data cleaning, extreme values in important numerical variables are examined using the IQR method.

```python
numerical_columns = [
    "BHK",
    "Size_in_SqFt",
    "Price_in_Lakhs",
    "Price_per_SqFt",
    "Year_Built",
    "Floor_No",
    "Total_Floors",
    "Age_of_Property",
    "Nearby_Schools",
    "Nearby_Hospitals"
]

outlier_summary = []

for col in numerical_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": outliers,
        "Outlier_Percentage": (outliers / len(df)) * 100
    })

outlier_summary_df = pd.DataFrame(outlier_summary)

outlier_summary_df
```

## 17. Phase 2 — Dataset Inspection Findings

The dataset contains 250,000 records and 23 columns.

### Data Quality Findings

- No missing values were found.
- No duplicate records were found.
- All IDs are unique.
- No invalid or negative values were found for major numerical fields.
- `Age_of_Property` is fully consistent with `Year_Built`.
- `Floor_No` contains no negative values, but 116,304 records have `Floor_No > Total_Floors`, indicating a structural inconsistency in the source dataset.
- `Price_per_SqFt` is inconsistent with `Price_in_Lakhs` and `Size_in_SqFt`.
- The raw `Price_per_SqFt` column will therefore not be directly relied upon for future analysis.
- No major statistical outliers were detected in the main numerical variables using the IQR method, except for the already identified `Price_per_SqFt` issue.
- The `Amenities` column contains multiple amenities in comma-separated form and will require feature engineering.
- Categorical variables are generally well distributed.
- `Locality` contains synthetic-looking locality identifiers and will be retained as provided by the source dataset.

### Key Issues Requiring Attention

1. `Price_per_SqFt` requires recalculation during feature engineering.
2. `Floor_No` and `Total_Floors` have a substantial structural inconsistency and should not be artificially corrected.
3. The dataset should be preserved in its raw form, while cleaning and transformations should be performed on a separate processed dataset.

### Conclusion

The dataset is largely complete and structurally usable, but several domain-level inconsistencies require careful handling during data cleaning and feature engineering.
